```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Superstore Sales Analysis", fontsize=16)

# 1. Category wise total sales
cat_sales = df.groupby("category")["sales"].sum()
axes[0, 0].bar(cat_sales.index, cat_sales.values, 
               color=["blue", "green", "red", "orange"])
axes[0, 0].set_title("Category wise Total Sales")
axes[0, 0].set_xlabel("Category")
axes[0, 0].set_ylabel("Sales")

# 2. Region wise total sales
reg_sales = df.groupby("region")["sales"].sum()
axes[0, 1].bar(reg_sales.index, reg_sales.values,
               color=["purple", "cyan", "brown", "pink"])
axes[0, 1].set_title("Region wise Total Sales")
axes[0, 1].set_xlabel("Region")
axes[0, 1].set_ylabel("Sales")

# 3. monthly sales trend
month_order = ["Jan","Feb","Mar","Apr","May",
               "Jun","Jul","Aug","Sep","Oct"]
month_sales = df.groupby("month")["sales"].sum()
month_sales = month_sales.reindex(month_order)
axes[1, 0].plot(month_sales.index, month_sales.values, 
                marker="o", color="blue")
axes[1, 0].set_title("Monthly Sales Trend")
axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("Sales")
axes[1, 0].grid(True)

# 4. Category wise Sales — Pie Chart
axes[1, 1].pie(cat_sales.values, labels=cat_sales.index, 
               autopct="%1.1f%%",
               colors=["blue", "green", "red", "orange"])
axes[1, 1].set_title("Category Sales Distribution")

plt.tight_layout()
plt.show()
```
